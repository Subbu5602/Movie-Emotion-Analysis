```python
df_clean = df[["title","year","imdb_rating","plot_feelings","f1_hope","f1_despair","f1_enlightenment","f1_resentment","f1_happiness","f1_sadness","f1_compassion","f1_disgust"]]
df_clean.head(30)
```

```python
df_clean = df_clean.dropna()
print(df_clean.shape)
```

```python
df_clean.describe()
```

```python
plt.figure(figsize=(17,15))
sns.histplot(df_clean["imdb_rating"],bins=40,kde=True,color="green")
plt.title("Distribution of IMDB Ratings for movies")
plt.xlabel("Ratings")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

#Trends between 7 and 8 usually
```

```python
#finding mean of all emotions and putting in ascending order
emotions = ["f1_hope","f1_despair","f1_enlightenment","f1_resentment","f1_happiness","f1_sadness","f1_compassion","f1_disgust"]
df_emotions_series = df_clean[emotions].mean().sort_values(ascending=False)
```

```python
df_emotions_series
```

```python
plt.figure(figsize=(10,11))
sns.barplot(x=df_emotions_series.index,y=df_emotions_series.values,palette="pink")
plt.title("Average Ratings of emotions invoked by movies")
plt.xlabel("Emotion")
plt.ylabel("Mean Rating")
plt.tight_layout()
plt.show()
```

```python
#classifying ratings in buckets, then checking emotion trends in each bucket
df_clean["ratings_group"] = pd.cut(df_clean["imdb_rating"],bins=[0,69,79,89,100],labels=["Below 70","70-80","80-90","90+"])
group_emotions = df_clean.groupby("ratings_group")[emotions].mean()

plt.figure(figsize=(14,14))
group_emotions.plot(kind="bar",figsize=(10,11),colormap="Set1")
plt.title("Emotions Present by Ratings")
plt.xlabel("Movie Rating Group")
plt.ylabel("Average Emotion Score")
plt.legend(bbox_to_anchor=(1.05,1),loc="upper right")
plt.tight_layout()
plt.show()
```

```python
#Checking correlation between emotions with ratings
corr_cols = ["imdb_rating","f1_hope","f1_despair","f1_enlightenment","f1_resentment","f1_happiness","f1_sadness","f1_compassion","f1_disgust"]
corr_matrix = df_clean[corr_cols].corr()

plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix,annot=True,fmt=".2f",cmap="coolwarm",center=0)
plt.title("Correlation Heatmap - Emotions and Rating")
plt.tight_layout()
plt.show()
```

```python
#Filtering into 2 buckets for high and low hope
avg_hope = df_clean["f1_hope"].mean()
df_clean["hope_group"] = df_clean["f1_hope"].apply(lambda x:"Higher Hope" if x>avg_hope else "Lower Hope")
```

```python
print(avg_hope)
```

```python
df_clean["hope_group"].value_counts()
```

```python
#checking if median ratings of films change with higher vs lower hopes
plt.figure(figsize=(8,6))
sns.boxplot(data=df_clean, x="hope_group", y="imdb_rating", palette="coolwarm")
plt.title("IMDB Ratings - High Hope vs Low Hope Films")
plt.xlabel("Hope in Films")
plt.ylabel("IMDB Rating")
plt.tight_layout()
plt.show()
```

```python
#checking no.of films for each decade

df_clean["decade"] = (df_clean["year"] // 10) * 10
df_clean["decade"].value_counts().sort_index()
```

```python
#filtering between 1950 and 2020 ad they have more data

df_decade = df_clean[(df_clean["decade"] >= 1950) & (df_clean["decade"] <= 2020)]
df_decade["decade"].value_counts().sort_index()
```

```python
decade_stats = df_decade.groupby("decade")[["imdb_rating"] + emotions].mean()
decade_stats["imdb_rating"].round(2)
```

```python
#top 5 emotions on avg by decade
top5_emotions_by_decade = decade_stats[emotions].apply(lambda row:row.nlargest(5).index.tolist(), axis=1)
top5_emotions_by_decade
```

```python
#General plotting by each genre per decade

decade_emotions = decade_stats[emotions]

decade_emotions.plot(kind="bar",figsize=(14,7),colormap="tab10")
plt.title("Top Emotions by Decade")
plt.xlabel("Decade")
plt.ylabel("Average Emotion Score")
plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.05,1),loc="upper left")
plt.tight_layout()
plt.show()
```

```python
#Top 4 emotions per decade

top4_per_decade = decade_stats[emotions].apply(lambda row:row.nlargest(4),axis=1)

top4_per_decade.plot(kind="bar",figsize=(14,7),colormap="tab10")
plt.title("Top 4 Emotions by Decade")
plt.xlabel("Decade")
plt.ylabel("Average Emotion Score")
plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.05,1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
#Trends for Happiness,Sadness,Compassion,Resentment

soft_emotions = ["f1_happiness","f1_sadness","f1_compassion","f1_resentment"]
decade_stats[soft_emotions].plot(kind="bar",figsize=(14,7),colormap="Set2")
plt.title("Soft Emotions by Decade")
plt.xlabel("Decade")
plt.ylabel("Average Emotion Score")
plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.05,1),loc="upper left")
plt.tight_layout()
plt.show()
```
