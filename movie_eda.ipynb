```python
#Checking correlation between emotions with ratings
corr_cols = ["imdb_rating","f1_hope","f1_despair","f1_enlightenment","f1_resentment","f1_happiness","f1_sadness","f1_compassion","f1_disgust"]
corr_matrix = df_clean[corr_cols].corr()

clean_cols = ["IMDB Rating","Hope","Despair","Enlightenment","Resentment","Happiness","Sadness","Compassion","Disgust"]
plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix,annot=True,fmt=".2f",cmap="coolwarm",center=0,xticklabels=clean_cols,yticklabels=clean_cols)
plt.title("Correlation Heatmap between Emotions and Ratings",fontsize=14)
plt.xticks(rotation=45,ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig("visuals/Correlation Heatmap.png",dpi=150,bbox_inches="tight")
plt.show()
```

```python
#checking no.of films for each decade

df_clean["decade"] = (df_clean["year"] // 10) * 10
df_clean["decade"].value_counts().sort_index()
```

```python
#filtering between 1950 and 2020 ad they have more data

df_decade = df_clean[(df_clean["decade"] >= 1950) & (df_clean["decade"] <= 2020)]
df_decade["decade"].value_counts().sort_index()
```

```python
decade_stats = df_decade.groupby("decade")[["imdb_rating"] + emotions].mean()
decade_stats["imdb_rating"].round(2)
```

```python
#top 5 emotions on avg by decade
top5_emotions_by_decade = decade_stats[emotions].apply(lambda row:row.nlargest(5).index.tolist(), axis=1)
top5_emotions_by_decade
```

```python
#Plottting all emotions for each decade
```

```python
#General plotting by each genre per decade

decade_emotions = decade_stats[emotions]

decade_emotions.plot(kind="bar",figsize=(14,7),colormap="tab10")
plt.title("Top Emotions by Decade")
plt.xlabel("Decade")
plt.ylabel("Average Emotion Score")
plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.05,1),loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Plotting top 4 emotions per decade
```

```python
from collections import Counter

all_feelings = df_clean["plot_feelings"].str.split(",").explode().str.strip().str.lower()
feeling_count = Counter(all_feelings)
feeling_count_df = pd.DataFrame(feeling_count.most_common(15),columns=["feeling","count"])

feeling_count_df
```

```python
df_clean["dominant_feeling"] = df_clean["plot_feelings"].str.split(",").str[0].str.strip().str.lower()
df_clean[["title","plot_feelings","dominant_feeling"]].head(30)
```

```python
# Pivot Table with Plot Feelings against Movie Ratings and film count - then plotting on barplot
# We are also dropping feelings that only appear dominant in less than 50 films
```

```python
pivot_feelings = df_clean.groupby("dominant_feeling")["imdb_rating"].agg(["mean","count"]).round(2)
pivot_feelings.columns = ["avg_rating","film_count"]
pivot_feelings = pivot_feelings[pivot_feelings["film_count"]>=50].sort_values("avg_rating",ascending=False)
pivot_feelings.head(30)
```

```python
plt.figure(figsize=(10,7))
sns.barplot(data=pivot_feelings,x="avg_rating",y="dominant_feeling",palette="coolwarm")
plt.title("Average IMDB Rating for each dominant plot feeling")
plt.xlabel("Average IMDB Rating")
plt.ylabel("Dominant Plot Feeling")
plt.tight_layout()
plt.savefig("visuals/Average Rating by Major Plot Feeling")
plt.show()
```

```python
# Cleaning Metascores data
```

```python
df[["title","imdb_rating","metascore"]].describe()
```

```python
df_critic = df_clean.copy()
df_critic["metascore"] = df["metascore"]
df_critic = df_critic.dropna(subset=["metascore"])
print(df_critic.shape)
```

```python
# Finding both Audience and Critic Ratings Correlation with Emotion 
audience_corr = df_critic[emotions].corrwith(df_critic["imdb_rating"])
critic_corr = df_critic[emotions].corrwith(df_critic["metascore"])

comparison = pd.DataFrame({
    "Audience(IMDB Score) : " : audience_corr,
    "Critic(Metascore) : " : critic_corr
})

comparison
```

```python
# Finding out top correlation of emotions across both audience and critics

audience_top = audience_corr.sort_values(ascending=False).reset_index()
audience_top.columns=("Emotion","Audience Correlation")

critic_top = critic_corr.sort_values(ascending=False).reset_index()
critic_top.columns=("Emotion","Critic Correlation")

pd.concat([audience_top,critic_top],axis=1)
```

```python
# Side by side barplot for ratings correlation with each emotion for c
```

```python
plt.figure(figsize=(10,7))
x = range(len(emotions))
width = 0.35

clean_emotions = [e.replace("f1_","").capitalize() for e in emotions]

audience_vals = [audience_corr[e] for e in emotions]
critic_vals = [critic_corr[e] for e in emotions]

plt.barh([i for i in x],audience_vals,width,label="Audience(IMDB)",color="midnightblue")
plt.barh([i + width for i in x],critic_vals,width,label="Critics(MetaScore)",color="maroon")

plt.yticks([i+width/2 for i in x],clean_emotions)
plt.xlabel("Correlation with Ratings")
plt.title("Emotion Valued between Audience and Critic")
plt.legend()
plt.tight_layout()
plt.savefig("visuals/Critics VS Audience.png",dpi=150,bbox_inches="tight")
plt.show()
```
