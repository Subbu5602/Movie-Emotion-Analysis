```python
df_clean = df[["title","year","imdb_rating","plot_feelings","f1_hope","f1_despair","f1_enlightenment","f1_resentment","f1_happiness","f1_sadness","f1_compassion","f1_disgust"]]
df_clean.head(30)
```

```python
df_clean = df_clean.dropna()
print(df_clean.shape)
```

```python
df_clean.describe()
```

```python
plt.figure(figsize=(17,15))
sns.histplot(df_clean["imdb_rating"],bins=40,kde=True,color="green")
plt.title("Distribution of IMDB Ratings for movies")
plt.xlabel("Ratings")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
emotions = ["f1_hope","f1_despair","f1_enlightenment","f1_resentment","f1_happiness","f1_sadness","f1_compassion","f1_disgust"]
df_emotions_series = df_clean[emotions].mean().sort_values(ascending=False)
```

```python
df_emotions_series
```

```python
plt.figure(figsize=(10,11))
sns.barplot(x=df_emotions_series.index,y=df_emotions_series.values,palette="pink")
plt.title("Average Ratings of emotions invoked by movies")
plt.xlabel("Emotion")
plt.ylabel("Mean Rating")
plt.tight_layout()
plt.show()
```

```python
df_clean["ratings_group"] = pd.cut(df_clean["imdb_rating"],bins=[0,69,79,89,100],labels=["Below 70","70-80","80-90","90+"])
group_emotions = df_clean.groupby("ratings_group")[emotions].mean()

plt.figure(figsize=(10,11))
group_emotions.plot(kind="bar",figsize=(10,11),colormap="Set1")
plt.title("Emotions Present by Ratings")
plt.xlabel("Movie Rating Group")
plt.ylabel("Average Emotion Score")
plt.legend(bbox_to_anchor(1.05,1),loc="upper right")
plt.tight_layout()
plt.show()
```
